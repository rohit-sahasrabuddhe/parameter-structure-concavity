This notebook contains the counterexample to Theorem 2 of [Ricotta and Szeidl (2006)](https://www.sciencedirect.com/science/article/abs/pii/S004058090600075X). We will show that a PD similarity matrix derived from a metric does not make entropy concave for alpha > 2.

```python
D = np.array(
    [[0.0, .72, 1.0],
     [.72, 0.0, .30],
     [1.0, .30, 0.0]]
)

Z = 1.0 - D
```

Z is positive definite:

```python
np.linalg.eigvalsh(Z)
```

```python
p = np.array([0.98, 0.01, 0.01])

hessian = get_hessian(Z, 6, p)
P = np.eye(3) - np.ones((3,3)) / 3
```

Hessian is not CNSD:

```python
is_CND(hessian, allow_semidefinite=True)
```

The projected Hessian has a positive eigenvalue:

```python
eigvals, eigvecs = np.linalg.eigh(P @ hessian @ P)
np.round(eigvals, 4)
```

```python
# direction
delta = np.round(eigvecs[:, -1], 2)

# magnitude
eps = 0.01
```

```python
np.round(
    ( get_entropy(Z, 6, p + eps*delta) + get_entropy(Z, 6, p - eps*delta) ) / 2.,
    7
)
```

```python
np.round(get_entropy(Z, 6, p), 7)
```

To verify:

```python
get_entropy(Z, 6, p) < ( get_entropy(Z, 6, p + eps*delta) + get_entropy(Z, 6, p - eps*delta) ) / 2.
```
