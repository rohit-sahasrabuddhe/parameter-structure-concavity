```python
Z = np.array([
    [1.00, 0.05, 0.04, 0.41],
    [0.05, 1.00, 0.62, 0.02],
    [0.04, 0.62, 1.00, 0.57],
    [0.41, 0.02, 0.57, 1.00]
])

np.linalg.eigvalsh(Z)
```
